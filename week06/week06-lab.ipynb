### **Task 3.1: Create Multi-Feature Dataset (2 points)**

Run this code to create a heart disease dataset with multiple features.

```python
# Create heart disease dataset
np.random.seed(123)

n_patients = 300

# Features
age = np.random.randint(30, 80, n_patients)
cholesterol = np.random.normal(200, 40, n_patients)
blood_pressure = np.random.normal(120, 20, n_patients)
bmi = np.random.normal(27, 5, n_patients)

# Generate heart disease based on features
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

disease_prob = sigmoid(
    -8 + 0.05 * age + 0.015 * cholesterol + 0.02 * blood_pressure + 0.1 * bmi
)
heart_disease = (np.random.random(n_patients) < disease_prob).astype(int)

# Create DataFrame
heart_df = pd.DataFrame({
    'age': age,
    'cholesterol': cholesterol,
    'blood_pressure': blood_pressure,
    'bmi': bmi,
    'heart_disease': heart_disease
})

print("Heart Disease Dataset:")
print(heart_df.head(10))
print(f"\nDataset shape: {heart_df.shape}")
print(f"\nDisease prevalence: {heart_df['heart_disease'].mean():.1%}")
```
