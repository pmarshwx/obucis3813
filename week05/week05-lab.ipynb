## **Exercise 1: Data Preparation (10 points)**

We'll use a subset of the Ames Housing dataset, which contains information about house sales in Ames, Iowa. This is a popular dataset for practicing regression techniques because it has many features and realistic complexity.

Run the cell below to load and prepare the data.

```python
# Load Ames Housing dataset
ames = fetch_openml(name="house_prices", as_frame=True, parser='auto')

# Select numeric features only for this lab
numeric_features = ['LotArea', 'OverallQual', 'OverallCond', 'YearBuilt',
                    'YearRemodAdd', 'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF',
                    'TotalBsmtSF', '1stFlrSF', '2ndFlrSF', 'LowQualFinSF',
                    'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
                    'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd',
                    'Fireplaces', 'GarageCars', 'GarageArea', 'WoodDeckSF',
                    'OpenPorchSF', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch',
                    'PoolArea', 'MiscVal', 'MoSold', 'YrSold']

X_ames = ames.data[numeric_features].copy()
y_ames = ames.target.astype(float)

# Handle missing values by filling with median
X_ames = X_ames.fillna(X_ames.median())

print("Ames Housing Dataset Loaded Successfully!")
print(f"="*50)
print(f"Number of samples: {X_ames.shape[0]}")
print(f"Number of features: {X_ames.shape[1]}")
print(f"\nFeatures: {list(X_ames.columns)}")
print(f"\nTarget variable: SalePrice (house sale price in $)")
print(f"Target range: ${y_ames.min():,.0f} - ${y_ames.max():,.0f}")
X_ames.head()
```
